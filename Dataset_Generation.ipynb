## Import Libraries

```python
import pandas as pd
import random
import numpy as np
from faker import Faker
from datetime import timedelta

fake = Faker("en_IN")
random.seed(42)
np.random.seed(42)
Faker.seed(42)
```

## Generate Cities

```python
cities = [
    ("Delhi", "Delhi"),
    ("Mumbai", "Maharashtra"),
    ("Bengaluru", "Karnataka"),
    ("Hyderabad", "Telangana"),
    ("Chennai", "Tamil Nadu"),
    ("Kolkata", "West Bengal"),
    ("Pune", "Maharashtra"),
    ("Ahmedabad", "Gujarat"),
    ("Jaipur", "Rajasthan"),
    ("Lucknow", "Uttar Pradesh"),
    ("Chandigarh", "Chandigarh"),
    ("Indore", "Madhya Pradesh"),
    ("Bhopal", "Madhya Pradesh"),
    ("Surat", "Gujarat"),
    ("Nagpur", "Maharashtra")
]

cities_df = pd.DataFrame(
    cities,
    columns=["city_name", "state"]
)

cities_df.insert(0, "city_id", range(1, len(cities_df)+1))

cities_df.head()
```

## saving it 

```python
cities_df.to_csv("../data/cities.csv", index=False)

print("Cities CSV Saved Successfully!")
```

## Generate Customers

```python
customers = []

for i in range(1,1001):

    gender = random.choice(["Male","Female"])

    if gender=="Male":
        first = fake.first_name_male()
    else:
        first = fake.first_name_female()

    last = fake.last_name()

    customers.append([

        i,
        first,
        last,
        gender,
        random.randint(18,60),
        fake.unique.phone_number()[:10],
        fake.unique.email(),
        random.randint(1,15),
        fake.date_between("-4y","today")

    ])

customers_df = pd.DataFrame(customers,columns=[

"customer_id",
"first_name",
"last_name",
"gender",
"age",
"phone",
"email",
"city_id",
"join_date"

])

customers_df.head()
```

```python
customers_df.to_csv("../data/customers.csv",index=False)
```

## Restaurants

```python
restaurant_names = [

"Burger Hub",
"Pizza Point",
"Biryani House",
"Tandoori Flames",
"Punjabi Tadka",
"Chinese Express",
"Spice Villa",
"Food Junction",
"South Treat",
"Royal Kitchen"

]

cuisines = [

"North Indian",
"South Indian",
"Chinese",
"Italian",
"Mughlai",
"Fast Food",
"Desserts"

]

restaurants=[]

for i in range(1,151):

    restaurants.append([

        i,
        random.choice(restaurant_names)+" "+str(i),
        random.choice(cuisines),
        random.randint(1,15),
        round(random.uniform(3.5,5.0),1),
        random.randint(2008,2024)

    ])

restaurants_df=pd.DataFrame(restaurants,columns=[

"restaurant_id",
"restaurant_name",
"cuisine",
"city_id",
"rating",
"opening_year"

])

restaurants_df.to_csv("../data/restaurants.csv",index=False)
```

## Delivery partners

```python
partners=[]

vehicles=["Bike","Scooter","Cycle"]

for i in range(1,251):

    partners.append([

        i,
        fake.name(),
        random.choice(vehicles),
        random.randint(1,10),
        round(random.uniform(3.5,5.0),1)

    ])

partners_df=pd.DataFrame(partners,columns=[

"partner_id",
"partner_name",
"vehicle_type",
"experience_years",
"rating"

])

partners_df.to_csv("../data/delivery_partners.csv",index=False)
```

## orders

```python
statuses=["Delivered","Cancelled","Preparing"]

orders=[]

for i in range(1,10001):

    orders.append([

        i,
        random.randint(1,1000),
        random.randint(1,150),
        random.randint(1,250),
        fake.date_time_between("-2y","now"),
        random.randint(15,70),
        random.choices(

            statuses,

            weights=[85,5,10]

        )[0],
        round(random.uniform(150,1800),2)

    ])

orders_df=pd.DataFrame(orders,columns=[

"order_id",
"customer_id",
"restaurant_id",
"partner_id",
"order_date",
"delivery_time_minutes",
"order_status",
"total_amount"

])

orders_df.to_csv("../data/orders.csv",index=False)
```

## order items

```python
foods = [
    ("Veg Burger","Fast Food"),
    ("Chicken Burger","Fast Food"),
    ("Paneer Pizza","Italian"),
    ("Chicken Pizza","Italian"),
    ("Veg Biryani","North Indian"),
    ("Chicken Biryani","North Indian"),
    ("Masala Dosa","South Indian"),
    ("Idli","South Indian"),
    ("Noodles","Chinese"),
    ("Fried Rice","Chinese"),
    ("Ice Cream","Desserts"),
    ("Brownie","Desserts")
]

items=[]

item_id=1

for order in range(1,10001):

    no_items=random.randint(1,4)

    for _ in range(no_items):

        food=random.choice(foods)

        items.append([

            item_id,
            order,
            food[0],
            food[1],
            random.randint(1,3),
            round(random.uniform(80,450),2)

        ])

        item_id+=1

items_df=pd.DataFrame(items,columns=[

"item_id",
"order_id",
"food_name",
"category",
"quantity",
"unit_price"

])

items_df.to_csv("../data/order_items.csv",index=False)
```

## payments

```python
payments=[]

methods=["UPI","Card","Cash","Wallet"]

for i in range(1,10001):

    payments.append([

        i,
        i,
        random.choice(methods),
        random.choices(

            ["Success","Failed"],

            weights=[95,5]

        )[0]

    ])

payments_df=pd.DataFrame(payments,columns=[

"payment_id",
"order_id",
"payment_method",
"payment_status"

])

payments_df.to_csv("../data/payments.csv",index=False)
```

## reviews

```python
feedbacks=[
    "Excellent food",
    "Very tasty",
    "Fast delivery",
    "Late delivery",
    "Packaging could improve",
    "Highly recommended",
    "Average experience",
    "Loved it",
    "Will order again",
    "Food was cold"
]

reviews=[]

for i in range(1,10001):

    reviews.append([

        i,
        i,
        random.randint(1,5),
        random.choice(feedbacks)

    ])

reviews_df=pd.DataFrame(reviews,columns=[

"review_id",
"order_id",
"customer_rating",
"feedback"

])

reviews_df.to_csv("../data/reviews.csv",index=False)
```
