```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+pymysql://root:0000@localhost/food_delivery_analytics"
)

print("Connected Successfully!")
```

```python
orders = pd.read_csv("../data/orders.csv")
order_items = pd.read_csv("../data/order_items.csv")
payments = pd.read_csv("../data/payments.csv")
reviews = pd.read_csv("../data/reviews.csv")
```

```python
orders.to_sql(
    "orders",
    engine,
    if_exists="append",
    index=False
)

print("Orders Imported!")
```

```python
orders.to_sql(
    "orders",
    engine,
    if_exists="append",
    index=False
)

print("✅ Orders Imported Successfully!")
```

```python
order_items.to_sql(
    "order_items",
    engine,
    if_exists="append",
    index=False
)

print("✅ Order Items Imported Successfully!")
```

```python
payments.to_sql(
    "payments",
    engine,
    if_exists="append",
    index=False
)

print("✅ Payments Imported Successfully!")
```

```python
reviews.to_sql(
    "reviews",
    engine,
    if_exists="append",
    index=False
)

print("✅ Reviews Imported Successfully!")
```

```python
reviews.to_sql(
    "reviews",
    engine,
    if_exists="append",
    index=False
)

print("✅ Reviews Imported Successfully!")
```
